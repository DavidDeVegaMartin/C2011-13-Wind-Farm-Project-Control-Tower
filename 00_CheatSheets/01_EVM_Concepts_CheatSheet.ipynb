# Conceptos básicos de Earned Value Management (EVM)

El Earned Value Management es un método para controlar un proyecto midiendo qué se había planificado, qué se ha hecho realmente y cuánto ha costado.

Los 3 pilares:



| Acrónimo | Nombre Completo | Significado (EVM) | Ejemplo Práctico en el Proyecto |
| :--- | :--- | :--- | :--- |
| **PV** | Planned Value | Lo que se debería haber hecho según el plan (y su valor en dinero) | Según el plan, a día de hoy deberíamos haber completado trabajo por valor de 100.000 € |
| **EV** | Earned Value | Lo que realmente se ha hecho (valorado según el presupuesto) | Hemos completado trabajo que según el presupuesto vale 80.000 € |
| **AC** | Actual Cost | Lo que realmente hemos gastado | Hemos gastado 95.000 € |
    

## Indicadores de rendimiento (cómo vamos)

| Acrónimo | Nombre Completo | Fórmula | Interpretación |
| :--- | :--- | :--- | :--- |
| **SV** | Schedule Variance | `EV – PV` | ¿Vamos adelantados o retrasados?<br>• **Positivo:** adelanto<br>• **Negativo:** retraso |
| **CV** | Cost Variance | `EV – AC` | ¿Estamos por debajo o por encima de presupuesto?<br>• **Positivo:** ahorro<br>• **Negativo:** sobrecoste |
| **SPI** | Schedule Performance Index | `EV / PV` | Eficiencia de plazo:<br>• **1.0:** según plan<br>• **&lt; 1.0:** más lento de lo previsto |
| **CPI** | Cost Performance Index | `EV / AC` | Eficiencia de coste:<br>• **1.0:** según presupuesto<br>• **&lt; 1.0:** más caro de lo previsto |

## Indicadores basados en tiempo (versión más moderna)

| Acrónimo | Nombre Completo | Qué aporta |
| :--- | :--- | :--- |
| **ES** | Earned Schedule | Equivalente temporal del Earned Value (en qué fecha deberíamos estar según el trabajo realizado) |
| **SPI(t)** | Schedule Performance Index (time) | Versión del SPI basada en tiempo en lugar de en dinero. Suele ser más fiable hacia el final del proyecto |
| **SV(t)** | Schedule Variance (time) | Desviación de plazo medida en tiempo |

## Forecasting (predicciones a futuro)

| Acrónimo | Nombre Completo | Qué significa |
| :--- | :--- | :--- |
| **BAC** | Budget at Completion | Presupuesto total planificado del proyecto (en nuestro caso ≈ 21,37 M€) |
| **EAC** | Estimate at Completion | Estimación de cuánto costará finalmente el proyecto |
| **ETC** | Estimate to Complete | Cuánto falta por gastar desde ahora hasta el final |
| **VAC** | Variance at Completion | Diferencia entre el presupuesto original (BAC) y lo que estimamos que costará al final (EAC) |

# Formas habituales de calcular el EAC

Hay varias maneras de estimar el coste final, según el “Performance Factor” (PF) que se use:

| Método | Lógica | Cuándo se usa |
| :--- | :--- | :--- |
| **PF = 1** | “A partir de ahora iremos según lo planificado” | Optimista / neutro |
| **PF = CPI** | “Seguiremos con la misma eficiencia de coste que hasta ahora” | Más realista si el sobrecoste es estructural |
| **PF = SPI o SCI** | Combina eficiencia de plazo y de coste | Cuando hay problemas tanto de tiempo como de dinero |

# Resumen visual muy simple

Planificado (PV)     →  Lo que debería estar hecho

Hecho (EV)           →  Lo que realmente está hecho

Gastado (AC)         →  Lo que realmente ha costado



SPI / CPI            →  ¿Cómo de bien lo estamos haciendo?

EAC                  →  ¿Cuánto costará al final?
