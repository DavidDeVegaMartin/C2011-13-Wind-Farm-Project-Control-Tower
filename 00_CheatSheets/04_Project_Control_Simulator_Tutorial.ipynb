# Tutorial: Construcción de un Simulador de Gestión de Proyecto

## Caso real – Parque Eólico C2011-13 (DSLIB)

**Objetivo de este tutorial**

Partiendo de los archivos originales de un proyecto real, vamos a construir de forma ordenada y comprensible un sistema de seguimiento y control de proyecto.

Este ejercicio está pensado para demostrar:
- Cómo se analiza un archivo de proyecto real
- Cómo se estructura la información para poder gestionarla
- Cómo se prepara un entorno de seguimiento semanal (simulador de gestión)

**Archivos de partida:**
- `C2011-13 Wind Farm.xlsx` → Datos principales del proyecto
- `C2011-13 Wind Farm.p2x` → Archivo de ProTrack
- `C2011-13 Wind Farm.pdf` → Ficha resumen del proyecto
