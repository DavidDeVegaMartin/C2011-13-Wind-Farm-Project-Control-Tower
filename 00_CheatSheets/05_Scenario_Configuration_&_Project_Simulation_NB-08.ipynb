# CHEAT SHEET — NB_08 Scenario Configuration & Project Simulation
## Preparación conceptual antes del desarrollo · Aplicación a MEINS

## 1. ¿Qué es un Project Scenario?
Un escenario es una representación hipotética de lo que sucedería en un proyecto si cambiase alguna de sus condiciones iniciales.

Partimos de un plan aprobado (Baseline) y modificamos uno o varios parámetros para analizar sus consecuencias.

Ejemplo:
- Baseline: una actividad tiene una duración planificada de 20 días.
- Escenario: se produce una incidencia que añade 7 días.
- Simulación: recalculamos el calendario.
- Resultado: comprobamos si cambia la fecha final, la ruta crítica o las holguras.
La pregunta fundamental es:

¿Qué ocurriría con el proyecto si sucediese X?

## 2. Conceptos fundamentales que debemos manejar

### 2.1. Baseline (línea base)
Es la referencia aprobada contra la que medimos las desviaciones.

En nuestro proyecto tenemos:

Indicador	Baseline

Actividades	135

Dependencias	150

Duración CPM	299 días

Actividades críticas	3

Alertas de suministro	38

Buffers excedidos	9

La baseline no se modifica cuando simulamos escenarios. Es nuestro punto de comparación.

### 2.2. Schedule Variance (desviación del calendario)
Es la diferencia entre el calendario original y el calendario resultante.

En nuestro simulador utilizaremos principalmente:
\[ SV_{days}=D_{scenario}-D_{baseline} \]

Ejemplo:

Indicador	Días

Baseline	299

Escenario	306

Desviación	+7


El proyecto necesitaría siete días adicionales respecto al plan original, según el calendario simulado.

Importante: esto es una desviación en días del cronograma. No es el Schedule Variance monetario de Earned Value Management.

### 2.3. Activity Duration (duración de actividad)

Tiempo necesario para completar una actividad.

Es importante distinguir:
- Duración planificada.
- Duración modificada por una incidencia.
- Duración real observada.
En nuestro caso solo tenemos la primera. Las otras dos serán valores simulados, nunca datos históricos reales.

### 2.4. Dependencies (dependencias)

Definen las relaciones temporales entre actividades.

Las cuatro relaciones habituales son:
    
Relación	Significado	Ejemplo

FS	Finish to Start	B comienza cuando termina A

SS	Start to Start	B comienza cuando comienza A

FF	Finish to Finish	B termina cuando termina A

SF	Start to Finish	B termina cuando comienza A


También pueden existir desfases (lags) entre actividades.

Esto es importante porque un retraso no siempre afecta únicamente a la actividad donde se produce: puede propagarse a sus sucesoras.

### 2.5. Critical Path (ruta crítica)
Es la secuencia de actividades que determina la duración mínima del proyecto bajo las condiciones del modelo.

Una actividad crítica tiene holgura total igual a cero.

Si su duración aumenta y no existen mecanismos de recuperación, puede aumentar la duración global del proyecto.

Pero una actividad no crítica también puede convertirse en crítica cuando consume su holgura.

### 2.6. Total Float (holgura total)
Es el margen de tiempo que puede retrasarse una actividad sin retrasar la fecha final del proyecto, manteniendo las condiciones del calendario calculado.

Ejemplo:
- Actividad A: 0 días de holgura.
- Actividad B: 30 días de holgura.
Un retraso de 5 días en B podría absorberse dentro de su margen, siempre que las relaciones y restricciones lo permitan.

No significa que siempre podamos disponer libremente de esos 30 días: otras actividades pueden compartir esa holgura.

## 3. Material Management: el concepto más importante para MEINS
Aquí tenemos que distinguir tres magnitudes que a menudo se confunden.

Concepto	Qué representa

Lead Time	Tiempo que necesita el proveedor para suministrar el material

Supply Delay	Retraso del suministro respecto a la fecha prevista

Activity Delay	Retraso que finalmente experimenta la actividad por falta de material


No son equivalentes.
Ejemplo: suministro de un componente

1. Planificación
   
   Material previsto para el día 100.
2. Incidencia del proveedor
   
   El suministro llega 12 días tarde: día 112.
3. Situación de la actividad
   
   Si existe stock disponible, el retraso puede no afectar a la actividad. Si no existe alternativa, puede retrasar su inicio.
4. Impacto en el proyecto
   
   Dependerá de las dependencias, la holgura y la posibilidad de reorganizar el trabajo.

Esto es precisamente lo que queremos demostrar: que no basta con detectar que un proveedor llega tarde. Hay que determinar las consecuencias operativas.

### 3.1. Buffer

Es un margen de protección frente a incertidumbres o retrasos.

En nuestro proyecto ya tenemos buffers asociados a los suministros.

Por ejemplo:
- Retraso: 10 días.
  
- Buffer: 14 días.

- Exceso: -4 días.

El buffer todavía no se ha consumido completamente.

Si el retraso es de 19 días, el exceso es de 5 días.

Pero superar el buffer no implica automáticamente que la actividad se retrase cinco días: debemos comprobar cómo se relacionan suministro y actividad.

## 4. Corrective Actions: pasar de detectar a gestionar
Esta es la competencia que más nos interesa demostrar.

Un Project Planner no se limita a informar de que existe una desviación. Debe analizar alternativas y facilitar decisiones.

Algunas medidas habituales:

| Medida | Objetivo | Posible contrapartida |
| :--- | :--- | :--- |
| **Expedite delivery** | Acelerar el suministro | Mayor coste logístico |
| **Alternative supplier** | Buscar otro proveedor | Riesgo de homologación o calidad |
| **Resequencing** | Cambiar el orden de actividades | Nuevas restricciones |
| **Crashing** | Reducir duración de una actividad | Más recursos y coste |
| **Parallel execution** | Ejecutar actividades simultáneamente | Mayor coordinación y riesgo |

Dos conceptos especialmente importantes:

Crashing: reducir la duración de una actividad mediante recursos adicionales, turnos extraordinarios u otros medios.

Fast Tracking: solapar actividades que inicialmente estaban planificadas de forma secuencial, aceptando un posible incremento de riesgo.

No son medidas intercambiables y no siempre son técnicamente viables.

## 5 Indicadores que necesitamos conocer
No vamos a crear decenas de KPIs. Estos son suficientes para el MVP:

| Indicador | Interpretación |
| :--- | :--- |
| **Project Duration** | Duración total calculada |
| **Forecast Finish** | Fecha final prevista del escenario |
| **Schedule Variance (days)** | Diferencia respecto al baseline |
| **Critical Activities** | Actividades con holgura cero |
| **Critical Path** | Secuencia crítica resultante |
| **Float Consumption** | Holgura consumida respecto al baseline |
| **Recovery Days** | Días recuperados mediante acciones |
| **Additional Cost** | Coste adicional estimado de la intervención |
    

## 6 Decisiones metodológicas para nuestro proyecto

Hay algunas reglas que quiero que tengamos claras antes de programar:

1. El baseline es inmutable. Cada escenario parte de una copia independiente.

2. Los retrasos de suministro no se convierten automáticamente en retrasos de actividad. Debemos modelar la relación entre ambos.

3. Las dependencias se respetan. No podemos acelerar una actividad ignorando sus restricciones técnicas.

4. Cada acción correctiva tiene que tener una hipótesis operativa. No reducimos duraciones arbitrariamente.

5. Los costes simulados deben tener una base. Si no tenemos tarifas de aceleración, podemos utilizar supuestos explícitos, pero no presentarlos como costes históricos.

6. Los resultados se comparan siempre con la baseline.
