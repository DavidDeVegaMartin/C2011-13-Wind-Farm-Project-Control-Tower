# 00_Chuleta Earned Value Management (EVM)

## 1. Los 3 datos básicos

| Sigla | Significado | Pregunta que responde |
| :--- | :--- | :--- |
| **PV** | Planned Value | ¿Cuánto trabajo debería estar hecho según el plan? |
| **EV** | Earned Value | ¿Cuánto trabajo está realmente hecho? |
| **AC** | Actual Cost | ¿Cuánto dinero hemos gastado realmente? |

## 2. ¿Cómo vamos? (Indicadores de rendimiento)

| Sigla | Fórmula | Qué te dice | Bueno si… |
| :--- | :--- | :--- | :--- |
| **SV** | EV – PV | ¿Adelantados o retrasados? | > 0 |
| **CV** | EV – AC | ¿Ahorro o sobrecoste? | > 0 |
| **SPI** | EV / PV | Eficiencia de plazo | ≥ 1.0 |
| **CPI** | EV / AC | Eficiencia de coste | ≥ 1.0 |


Regla rápida:

* SPI o CPI = 1.0 → Vamos según lo previsto

* SPI o CPI < 1.0 → Vamos mal (retraso o sobrecoste)

* SPI o CPI > 1.0 → Vamos mejor de lo previsto

## 3. Versión temporal (más fiable al final del proyecto)

| Sigla | Significado |
| :--- | :--- |
| **ES** | Earned Schedule (versión en tiempo del EV) |
| **SPI(t)** | Eficiencia de plazo medida en tiempo |
| **SV(t)** | Desviación de plazo medida en tiempo |

## 4. ¿Cuánto costará al final? (Forecasting)

| Sigla | Significado |
| :--- | :--- |
| **BAC** | Budget at Completion → Presupuesto total planificado |
| **EAC** | Estimate at Completion → Estimación del coste final |
| **ETC** | Estimate to Complete → Cuánto falta por gastar |
| **VAC** | Variance at Completion → Desviación final esperada (BAC – EAC) |

## 5. Formas comunes de calcular el EAC

| Método | Fórmula orientativa | Mentalidad |
| :--- | :--- | :--- |
| **PF = 1** | `AC + (BAC – EV)` | “A partir de ahora iremos según el plan” |
| **PF = CPI** | `BAC / CPI` | “Seguiremos con la misma eficiencia de coste” |
| **PF = SPI o SCI** | Combina plazo + coste | Cuando hay problemas de tiempo y de dinero |

## 6. Lectura rápida de un proyecto

| Situación | SPI | CPI | Interpretación |
| :--- | :--- | :--- | :--- |
| **Ideal** | ≈ 1.0 | ≈ 1.0 | Según plan |
| **Retraso** | < 1.0 | ≈ 1.0 | Vamos tarde, pero el coste está controlado |
| **Sobrecoste** | ≈ 1.0 | < 1.0 | Vamos a tiempo, pero gastando de más |
| **Problema doble** | < 1.0 | < 1.0 | Retraso + sobrecoste |

* Al final del proyecto:
