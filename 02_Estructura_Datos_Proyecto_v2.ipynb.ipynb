# 02 – Estructura de Datos del Proyecto

**Objetivo de este notebook**

Definir de forma clara y ordenada **qué información necesitamos** y **cómo la vamos a estructurar** para construir un sistema de seguimiento de proyecto completo y usable.

Partimos de lo descubierto en el Notebook 01 y diseñamos el modelo de datos antes de transformar nada.

## Resumen de lo descubierto en el Notebook 01

- Existe una planificación base bien definida (`Baseline Schedule`)
- Hay **120 periodos de seguimiento** reales (`TP1` a `TP120`)
- Existe una hoja resumen de rendimiento (`Tracking Overview`)
- El proyecto tiene aproximadamente **167 actividades**
- Duración planificada: **525 días**
- Presupuesto planificado (BAC): **≈ 21,37 M€**

El archivo original es completo, pero no está pensado para usarse como herramienta de gestión diaria.

## ¿Qué información necesitamos para gestionar el proyecto?

Para poder **simular la gestión del proyecto semana a semana**, necesitamos tres niveles de información:

### 1. Definición estable de las actividades
Lo que no cambia (o cambia muy poco):
- Identificador, nombre, WBS
- Predecesoras y sucesoras
- Duración y coste planificados
- Fechas baseline

### 2. Seguimiento global del proyecto (Nivel 1)
Estado del proyecto completo en cada periodo de control:
- PV, EV, AC
- SPI, CPI, SPI(t)
- Variaciones de plazo y coste
- Fechas de cada control

### 3. Seguimiento detallado por actividad (Nivel 2)
Estado de **cada actividad** en cada periodo de control:
- Avance real
- Coste real acumulado
- Duración real
- Desviaciones respecto al plan

Este tercer nivel es el que permite un seguimiento fino y realista.

## Modelo de datos propuesto

Vamos a trabajar con **tres tablas principales**:

---

### Tabla 1: Activity Master
**Granularidad:** 1 fila = 1 actividad

| Campo | Descripción |
|-------|-------------|
| ID | Identificador único |
| WBS | Código de estructura de desglose |
| Name | Nombre de la actividad |
| Predecessors | Actividades predecesoras |
| Successors | Actividades sucesoras |
| Baseline_Start | Fecha de inicio planificada |
| Baseline_End | Fecha de fin planificada |
| Duration_Plan | Duración planificada |
| Total_Cost_Plan | Coste planificado |
| Phase | Fase del proyecto |
| Is_Critical_Material | Flag de material / equipo crítico |

---

### Tabla 2: Project Tracking (Nivel 1)
**Granularidad:** 1 fila = 1 Tracking Period

| Campo | Descripción |
|-------|-------------|
| TP | Número del periodo de seguimiento |
| Status_Date | Fecha de control |
| PV | Planned Value |
| EV | Earned Value |
| AC | Actual Cost |
| SPI | Schedule Performance Index |
| CPI | Cost Performance Index |
| SPI_t | Schedule Performance Index (time) |
| SV | Schedule Variance |
| CV | Cost Variance |
| p_factor | p-factor |

---

### Tabla 3: Activity Tracking (Nivel 2)
**Granularidad:** 1 fila = 1 Actividad × 1 Tracking Period

| Campo | Descripción |
|-------|-------------|
| TP | Número del periodo de seguimiento |
| Status_Date | Fecha de control |
| Activity_ID | Identificador de la actividad |
| Actual_Start | Fecha real de inicio |
| Actual_Duration | Duración real |
| Actual_Cost | Coste real |
| Pct_Complete | Porcentaje de avance |
| EV | Earned Value de la actividad |
| PV | Planned Value de la actividad |

Esta tabla es la más potente para simular el seguimiento detallado del proyecto.

## Principios de diseño

1. **Separación clara** entre:
   - Definición del proyecto (Activity Master)
   - Seguimiento global (Project Tracking)
   - Seguimiento detallado (Activity Tracking)

2. **Formato consistente** a lo largo de todo el tiempo

3. **Nombres de campos claros** y profesionales

4. **Orientado a uso real**:
   - Se puede usar en Excel
   - Se puede conectar fácilmente a Power BI
   - Permite simular la gestión semana a semana

5. **Escalable**: primero construimos la base, después añadimos lógica de alarmas y cuadro de mando

## Conclusión

Hemos definido un modelo de datos completo y estructurado:

- **Activity Master** → Qué es el proyecto
- **Project Tracking** → Cómo evoluciona el proyecto en global
- **Activity Tracking** → Cómo evoluciona cada actividad en el tiempo

Este modelo nos permite construir un **simulador de gestión de proyecto** realista y profesional.
